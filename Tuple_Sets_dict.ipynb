```python
cities = {"Delhi","Agra"}
print(cities)
print(type(cities))
```

```python
cities.add("Mathura")
print(cities)
cities.update(["lucknow","Kanpur"])
print(cities)
```

```python
cities.remove("Agra")
```

```python
print(cities)
cities.discard("Luccknow")
print(cities)
```

```python
cities.discard("delhi")
print(cities)
```

```python
print(cities)
rem = cities.pop()
print(rem)
print(cities)
```
